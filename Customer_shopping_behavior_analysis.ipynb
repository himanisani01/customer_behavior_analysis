```python
#Replace Null values with Mean and Median because - Median > Mean because Mean can get affected by outliers and Median is more ROBUST
#will replace null value with Median review rating value of each category
#Ex-If a clothing review is missed so it will be replaced by a clothing item review only
```

```python
#First Group the data Second Select column Third Apply function and last transform() 
```

```python
df["Review Rating"]=df.groupby("Category")["Review Rating"].transform(lambda x: x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
df.columns = df.columns.str.lower() #Lower Case
df.columns = df.columns.str.replace(' ','_') #Replacing Space with Underscore
df = df.rename(columns={"purchase_amount_(usd)":"purchase_amount"})
```

```python
df.columns
```

```python
#Create a column age_group
labels=["Young Adult", "Adult", "Middle-Aged", "Senior"]
df["age_group"] = pd.qcut(df["age"], q=4, labels = labels)
```

```python
df[["age","age_group"]].head(10)
```

```python
#Create column Purchase frequency_days #Textual frequency into numbers
frequency_mapping = {
    'Fornightly':14,
    'Weekly':7,
    'Monthly':30,
    'Quarterly':90,
    'Bi-weekly':14,
    'Annually':365,
    'Every 3 months':90
}

df['purchase_frequency_days']=df['frequency_of_purchases'].map(frequency_mapping)
```
